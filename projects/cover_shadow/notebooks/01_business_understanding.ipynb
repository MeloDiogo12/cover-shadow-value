## 1. Problema

Quando se fala do momento defensivo, ou do momento sem bola, pensa-se habitualmente em recuperações de bola, interceções ou duelos. Contudo, importa considerar também o condicionamento do adversário, até porque há jogadores com um impacto defensivo considerável que quase não recuperam bolas, uma vez que o seu posicionamento e a sua orientação corporal, quando corretos, fecham linhas de passe e orientam o adversário para um determinado corredor.

Um exemplo ajuda a concretizar esta ideia. Considere-se uma equipa sem bola, organizada em bloco médio, com a bola num defesa central adversário. Nesse momento, o extremo tem, em termos gerais, duas opções: manter-se por fora, perto do lateral adversário, ou posicionar-se por dentro, fechando as linhas de passe interiores e orientando o corpo de forma a encaminhar a bola para fora, zona onde a equipa consegue pressionar de forma mais eficaz. Decisões desta natureza são centrais no trabalho de um treinador e pesam, muitas vezes, na qualidade com que uma equipa defende.

As métricas defensivas tradicionais (desarmes, interceções, entre outras), porém, não traduzem o valor destas ações, porque registam apenas o que acontece com a bola. Um extremo com poucos desarmes ou interceções pode, assim, ser visto como tendo pouca carga de trabalho defensivo e ser, ao mesmo tempo, fundamental na forma como a equipa defende.

Existem já trabalhos que procuram dar valor a este tipo de ações, como o de Cascioli et al. (2025), com dados StatsBomb 360. Contudo, nesse trabalho a sombra de cobertura é definida a partir da linha de passe e do alcance do defesa, em fotografias estáticas do momento do passe, sem considerar a orientação corporal, componente essa que o presente trabalho procura acrescentar, a par de alguns refinamentos na definição.

Para isso, será necessário definir com dados o que é uma sombra de cobertura, construída a partir do posicionamento e da orientação corporal do defesa, distingui-la da marcação individual, atribuir valor a uma linha de passe fechada por essa sombra e repartir esse valor quando mais do que um defesa contribui para o seu fecho. O âmbito do projeto limita-se, por isso, à demonstração de uma abordagem possível para quantificar esta ação, cuja componente corporal tem sido, até aqui, percetível sobretudo pela observação.

## 2. Objetivos

**Objetivo geral:** demonstrar como se pode medir, com tracking e body pose, o valor do
posicionamento defensivo sem bola através das sombras de cobertura, com o cone definido pela
orientação corporal do defesa.

**Objetivos específicos:**
- definir operacionalmente uma sombra de cobertura a partir de tracking e body pose;
- distinguir sombra de cobertura de marcação individual;
- atribuir valor a uma linha de passe fechada e reparti-lo entre os defesas que a fecham;
- avaliar a estabilidade da métrica e o contributo da orientação corporal;
- comunicar o método de forma explicável a treinadores e analistas.

**Perguntas de investigação:**
- PI1 · Como definir operacionalmente uma sombra de cobertura a partir de tracking e body pose?
- PI2 · Como atribuir valor a uma linha de passe fechada e repartir esse valor entre defesas?
- PI3 · A orientação corporal altera a sombra face a uma definição baseada apenas na posição relativa à bola?
- PI4 · A métrica é estável face a variações razoáveis dos seus parâmetros?

## 3. Critérios de sucesso

Não existe variável-alvo a prever (tal como no TFM); o sucesso avalia-se por estes critérios.

| # | Critério | Como se verifica |
|---|---|---|
| C1 | Validade aparente: a métrica vê o que um treinador vê | Revisão de 10–15 lances por um treinador, confirmando se os cones e as linhas fechadas fazem sentido |
| C2 | Separa sombra de marcação | Nesses lances, os defesas classificados como "a marcar" estão de facto a marcar |

| C6 | Reprodutível e rastreável | Notebooks correm do início ao fim; cada número volta aos dados; testes passam |
| C7 | Explicável (opcional) | 2–3 treinadores percebem a animação de um lance em menos de um minuto |

## 4. Avaliação da situação

**Recursos:** SkillCorner Open Data (tracking a 10 fps e eventos dinâmicos de 20 jogos da
A-League 2024/25; body pose a 25 fps de 2 jogos); ambiente (Python, visualização,
vídeo).

**Restrições:** body pose só em 2 jogos e só para jogadores detetados pela câmara; tracking
extrapolado fora da imagem da transmissão; uma única liga e época; sem vídeo da transmissão.

**Riscos:** erro na estimativa das articulações (erro p90 por articulação); desalinhamento entre
tracking e body pose (gerados separadamente); identidades erradas (~97% corretas, segundo a
SkillCorner); poucos momentos com todos os defesas relevantes com body pose.

**Pressupostos:** a métrica refere-se a passes rasteiros; o perigo de cada linha é estimado por
um modelo (A5 em `DECISOES.md`); o comprimento do cone representa o limite do passe por cima.

**Questões éticas e legais:** dados públicos com licença MIT; crédito à SkillCorner em todas as
figuras e vídeos; o objetivo é demonstrar um método, não avaliar jogadores concretos.
